## (a) 
Estimate the expected number of cycles to failure $E[N_1]$ and the variance $Var(N_1)$
for the safety limit. Do this by generating 100, 1,000, and 10,000 simulated values of N1.
Based on your simulations, what do you think the true values of $E[N_1]$ and $Var(N_1)$ are?

Calculate $N_1 = \min\{n : \sum_{i=1}^n U_i > 1\}$ by summing trials of a uniform random variable between [0,1] until it exceeds 1, then log the number of trials it took. Average those numbers across the trials.

```python
import random      
```

```python
def cyclesToN1Failure(trials):
#expectation
    cycles = []
    for trial in range(trials):
        total = 0
        cycle = 0
        while total <= 1:
            total += random.random()
            cycle += 1
        cycles.append(cycle)
    exp = sum(cycles) / len(cycles)

    #variance
    squared_diffs = []
    for x in cycles:
        diff = x - exp
        squared_diffs.append(diff ** 2)
    variance = sum(squared_diffs) / len(cycles)
    #print(f"Trials: {trials}, Expectation: {exp:.4f}, Variance: {variance:.4f}")
    return cycles

cyclesToN1Failure(100)
cyclesToN1Failure(1000)
cyclesToN1Failure(10000)
```

## (a) 
Let the profit per sale be $p=15$ and the hourly operating cost be $c=3$. Assume the vendor can only work an integer number of hours per day, up to a maximum of 10 hours. Design and code a simulation-based algorithm to find the optimal number of hours the vendor should work to maximize their expected net income.

```python
import numpy as np
import matplotlib.pyplot as plt

def income_list(hours, trials=10000):
    incomes = []
    for i in range(trials):
        sales = np.random.poisson(np.log(1 + hours))
        incomes.append(15 * sales - 3 * hours)
    return incomes

oneHr = income_list(1)
twoHr = income_list(2)
threeHr = income_list(3)
fourHr = income_list(4)
fiveHr = income_list(5)
sixHr = income_list(6)
sevenHr = income_list(7)
eightHr = income_list(8)
nineHr = income_list(9)
tenHr = income_list(10)

all_lists = [oneHr, twoHr, threeHr, fourHr, fiveHr, sixHr, sevenHr, eightHr, nineHr, tenHr]
averages = [sum(l) / len(l) for l in all_lists]
best = averages.index(max(averages)) + 1
print(f"Optimal hours: {best}")

plt.bar(range(1, 11), averages, edgecolor="black", color="lightgreen")
plt.xticks(range(1, 11))
plt.xlabel("Hours worked")
plt.ylabel("Average net income ($)")
plt.title("Expected net income by hours worked")
plt.show()
```
